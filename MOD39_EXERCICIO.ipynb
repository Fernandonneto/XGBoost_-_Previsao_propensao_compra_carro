# 3. Faça a matriz de correlação e analise as váriaveis que parecem ter maior correlação com a váriavel target.

```python
# Verificação da correlação entre as variáveis da base
var_correlation = base.corr()

# Exibição da matriz de correlação
plt.figure(figsize=(10, 8))
sns.heatmap(var_correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlação')
plt.show()
```

A análise da matriz de correlação aponta `Age` (**0.62**) e `AnnualSalary` (**0.36**) como as variáveis com maior associação linear positiva com o alvo `Purchased`. Em contrapartida, `Gender` apresenta uma correlação de **-0.05**, indicando uma relação linear praticamente nula (neutra) com a propensão de compra.
